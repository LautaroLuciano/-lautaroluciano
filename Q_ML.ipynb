```python
import random

n_vars = 10
terms = []

# Localidad 1: costos individuales ¿cREO?
for i in range(n_vars):
    coef = random.choice([0.5, 1.0, 1.5])  # costo de recorte
    label = ''.join('Z' if j == i else 'I' for j in range(n_vars))
    terms.append((label, coef))

# Localidad 2: pares
for _ in range(5):  # La cantidad de interacciones aleatorias amigín
    i, j = random.sample(range(n_vars), 2)
    coef = random.choice([0.3, 0.6])
    label = ''.join('Z' if k in (i,j) else 'I' for k in range(n_vars))
    terms.append((label, coef))

# Localidad 3: triples
for _ in range(3):
    i, j, k = random.sample(range(n_vars), 3)
    coef = random.choice([0.2, 0.4])
    label = ''.join('Z' if m in (i,j,k) else 'I' for m in range(n_vars))
    terms.append((label, coef))

print("Instancia generada con", n_vars, "variables y", len(terms), "términos.")
for t in terms:
    print(t)





```

```python
import itertools

def bit_energy(bitstr, terms):
#El salame del hamiltoniano calcula la energía del bitstring
    energy = 0.0
    for label, coef in terms:
        prod = 1
        for i, p in enumerate(label):
            if p == 'Z':
                # Convención: '0' -> +1, '1' -> -1
                prod *= (1 if bitstr[i] == '0' else -1)
        energy += coef * prod
    return energy

def brute_force_min_energy(n_vars, terms):
#Evalúa todos los estados posibles y devuelve el óptimo exacto
    best_energy = None
    best_state = None
    for bits in itertools.product('01', repeat=n_vars):
        bitstr = ''.join(bits)
        e = bit_energy(bitstr, terms)
        if best_energy is None or e < best_energy:
            best_energy = e
            best_state = bitstr
    return best_energy, best_state

# Si lo calculo sobre 10 variables:
n_vars = 10
best_energy, best_state = brute_force_min_energy(n_vars, terms)

print("Óptimo exacto:")  # O sea, cantidad mínima de energía posible. El OPTIMO EXACTO / MINIMIZA LA FUNCION DE COSTOOOOOO
print("Energía mínima =", best_energy)
print("Estado óptimo =", best_state)
```
